````python
#### 44. Consider a random 10x2 matrix representing cartesian coordinates, convert them to polar coordinates (★★☆)

# Create a 10x2 matrix with random Cartesian coordinates (x, y)
Z = np.random.random((10,2))

# Separate the x and y coordinates into individual arrays
X, Y = Z[:,0], Z[:,1] # X gets the first column, Y gets the second column

# Calculate the radial distance (R) using the Pythagorean theorem: R = sqrt(x^2 + y^2)
R = np.sqrt(X**2 + Y**2)

# Calculate the angular coordinate (T, or theta) using arctan2.
# arctan2 handles all four quadrants correctly, returning an angle in radians.
T = np.arctan2(Y, X)

# Print the calculated radial distances
print(R)
# Print the calculated angular coordinates (in radians)
print(T)
````
